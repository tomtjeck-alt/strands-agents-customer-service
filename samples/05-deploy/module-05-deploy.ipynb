## Part 1: The Deployment Code (`main.py`)

The only change from a local agent is the **entry point**. Locally you call `agent(prompt)` in a loop; on AgentCore you expose a handler that receives a `payload` and returns a response. The workshop's `main.py` wraps the same agent - same tools, same steering handlers, same context management (`context_manager="auto"`) - behind that handler.

```python
from bedrock_agentcore.runtime import BedrockAgentCoreApp

app = BedrockAgentCoreApp()

@app.entrypoint
def invoke(payload, context):
    prompt = payload.get("prompt")
    if not prompt:
        raise ValueError("Missing required field: prompt")
    agent = get_agent()
    response = agent(prompt)
    return str(response).strip()

if __name__ == "__main__":
    app.run()
```

- `BedrockAgentCoreApp()` creates the runtime app.
- `@app.entrypoint` marks the handler AgentCore calls on each invocation - it receives `payload` (a dict with the user prompt) and `context`.
- Returning `str(response).strip()` sends back the agent's reply as plain text.
- `app.run()` lets you run the same file locally as an HTTP server.

Open `main.py` in this folder to see the full agent (it imports `customer_service_tools.py` and `steering_handlers.py` - the same code from earlier modules).

## Part 3: Deploy with the AgentCore CLI

The rest of this module runs in a **terminal** from `/workshop/samples/05-deploy`. The `agentcore` CLI comes from the `@aws/agentcore` npm package.

### Step 1 - Move into the module folder and check the CLI

```bash
cd /workshop/samples/05-deploy
agentcore --version
```

Your agent code is already here: `main.py`, `customer_service_tools.py`, `steering_handlers.py`, `skills/`, and `requirements.txt`.

If `agentcore --version` prints `bash: agentcore: command not found`, install the CLI and check again:

```bash
sudo npm install -g @aws/agentcore@1.0.1 --ignore-scripts
agentcore --version
```

### Step 2 - Create the AgentCore project

```bash
agentcore create --name csdeploy --template empty
```

`--template empty` creates the project shell in a `csdeploy/` folder, with an `agentcore/` folder (config + CDK infrastructure). It does **not** generate agent code yet - you bring your own in the next step.

### Step 3 - Add your agent (Bring Your Own Code)

Move into the project folder that `create` made, then add a runtime for your agent:

```bash
cd csdeploy
agentcore add runtime --name MyAgent --language python
```

This registers a runtime named `MyAgent` with a direct code deploy (your code is zipped to S3 - no container) and `main.py` as the entrypoint. It also creates an `app/MyAgent/` folder with a `README.md` that describes what your code needs.

### Step 4 - Copy your agent code into `app/MyAgent/`

`add` created the `app/MyAgent/` folder, with no agent code yet. Copy the agent you built into it - `main.py` plus the files it imports:

```bash
cp ../main.py ../customer_service_tools.py ../steering_handlers.py ../requirements.txt app/MyAgent/
cp -r ../skills app/MyAgent/
```

The `entrypoint` (`main.py`) must sit at the root of that folder.

### Step 4b - Set up dependencies in `app/MyAgent/`

`agentcore deploy` builds the agent from a **`pyproject.toml`** in `app/MyAgent/` (a plain `requirements.txt` is not enough - the CDK build needs the `pyproject.toml`). Generate it with `uv` and add the dependencies:

```bash
cd app/MyAgent
uv init --bare --python 3.13
uv add strands-agents bedrock-agentcore aws-opentelemetry-distro boto3
cd ../..
```

`uv init --bare` creates the `pyproject.toml` without overwriting your `main.py`; `uv add` writes the dependencies.

> If you skip this, `agentcore deploy` fails with `Required project file not found: .../pyproject.toml`.

### Step 5 - Test locally with `agentcore dev` (optional)

```bash
agentcore dev
```

This starts your agent on a local development server (`http://127.0.0.1:8080`). Leave it running, open a second terminal in the same `csdeploy` folder, and send it a prompt:

```bash
agentcore invoke --local --runtime MyAgent --payload '{"prompt": "Hi, I am customer C-1001. What are my recent orders?"}'
```

Press **Ctrl+C** in the first terminal to stop the server.

### Step 6 - Deploy to AgentCore Runtime

```bash
agentcore deploy
```

`agentcore deploy` validates the project, synthesizes CloudFormation, zips your code to an Amazon S3 staging bucket, and provisions the AgentCore Runtime. The first deploy takes a few minutes (it installs dependencies); later updates reuse the cached dependencies and are faster.

### Step 7 - Invoke the deployed agent

```bash
agentcore invoke --runtime MyAgent --payload '{"prompt": "Hi, I am customer C-1001. What are my recent orders?"}'
```

`--runtime` is the runtime name from Step 3, and `--payload` is the JSON that `main.py` receives. Use `--session-id` to keep context across calls. The session ID must be **at least 33 characters** long (a shorter one fails with `Value at 'runtimeSessionId' failed to satisfy constraint: Member must have length greater than or equal to 33`):

```bash
agentcore invoke --runtime MyAgent --session-id customer-1001-session-orders-0000001 --payload '{"prompt": "Hi, I am customer C-1001. What are my recent orders?"}'
agentcore invoke --runtime MyAgent --session-id customer-1001-session-orders-0000001 --payload '{"prompt": "Which of those orders is still on its way?"}'
```

The second call answers without asking for your customer ID again, because both calls share the same session.

If a steering handler pauses `process_refund` for human approval, the reply is `{"status": "interrupted", "interrupts": [...]}` instead of text. Approve (`true`) or deny (`false`) it in the same session, with the `interruptId` from that reply:

```bash
agentcore invoke --runtime MyAgent --session-id customer-1001-session-orders-0000001 --payload '{"interrupt_responses": [{"interruptId": "<interruptId>", "response": true}]}'
```

### Step 8 - Invoke from code with boto3 (the production path)

`agentcore invoke` is for testing from the terminal. In production you call the deployed runtime directly with the **AWS SDK** (`InvokeAgentRuntime`). Same agent, two ways to reach it.

Get the runtime ARN with `agentcore status` - it prints something like `arn:aws:bedrock-agentcore:us-east-1:<account>:runtime/<project>_<name>-XXXXXXXXXX`. Paste it into the cell below and run it.

````python
import json
import uuid
import boto3

# Paste the ARN from `agentcore status`:
agent_arn = "<VOTRE_AGENT_ARN>"
prompt = "Hi, I'm customer C-1001. What are my recent orders?"

client = boto3.client("bedrock-agentcore", region_name="us-east-1")

response = client.invoke_agent_runtime(
    agentRuntimeArn=agent_arn,
    runtimeSessionId=str(uuid.uuid4()),  # must be 33+ chars; a uuid4 satisfies that
    payload=json.dumps({"prompt": prompt}).encode(),
    qualifier="DEFAULT",
)

# The response body streams back in chunks - join and decode them
chunks = [chunk.decode("utf-8") for chunk in response.get("response", [])]
print("".join(chunks))
````

## Troubleshooting

| Symptom | Resolution |
|---------|------------|
| `bash: agentcore: command not found` | The global npm install was left incomplete, so the `agentcore` command points at a file that does not exist. Reinstall skipping the package's `postinstall`: `sudo npm install -g @aws/agentcore@1.0.1 --ignore-scripts`, then confirm with `agentcore --version`. |
| `error: too many arguments for 'invoke'` or `error: unknown option '-y'` | The commands come from an older CLI version. Use the commands on this page, tested with AgentCore CLI 1.0.1 (`agentcore --version`). |
| `unknown command 'configure'` | You have the **old** CLI. `configure` was replaced - use `agentcore create` + `agentcore add`. Uninstall the old toolkit: `pip uninstall bedrock-agentcore-starter-toolkit`. |
| `agentcore deploy` fails with `Required project file not found: .../pyproject.toml` | The agent folder (`app/MyAgent/`) has no `pyproject.toml`. Run Step 4b: `cd app/MyAgent && uv init --bare --python 3.13 && uv add strands-agents bedrock-agentcore aws-opentelemetry-distro boto3`. |
| `npm error code EACCES` on `npm install -g` | A global npm install needs root. Use `sudo npm install -g @aws/agentcore@1.0.1 --ignore-scripts`. (In Workshop Studio it is already installed.) |
| `agentcore deploy` fails on region | The first deploy creates the deployment target in `agentcore/aws-targets.json` from your AWS account and region. Make sure its `region` is `us-east-1`. |
| `AccessDeniedException` on Bedrock | The first call to a model auto-initiates access in the background; wait a couple of minutes and retry. |
| `Value at 'runtimeSessionId' failed to satisfy constraint: Member must have length greater than or equal to 33` | The `--session-id` you passed to `agentcore invoke` is too short - AgentCore requires session IDs of **33+ characters**. Use a longer ID (e.g. `customer-1001-session-refund-000001`) or a UUID. |

> The Workshop Studio environment is already CDK-bootstrapped, so `agentcore deploy` works without any `cdk bootstrap` step. (If you run this locally instead, `agentcore deploy` bootstraps your account automatically on first use.)

## Cleanup

When you're done, tear down the AWS resources this module created. Cleanup is **two steps**: reset the project config, then deploy that empty state so AWS removes the resources.

```bash
agentcore remove all --yes    # clears the local config (does NOT touch AWS yet)
agentcore deploy --yes        # applies the teardown - removes the runtime from AWS
```

`agentcore remove all --yes` only resets `agentcore/agentcore.json` locally; the follow-up `agentcore deploy --yes` is what actually deletes the AgentCore Runtime and its CloudFormation stack. Without `--yes`, `deploy` stops and asks you to confirm the deletion.
